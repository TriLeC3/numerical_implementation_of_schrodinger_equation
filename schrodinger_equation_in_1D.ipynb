```python
import numpy as np
```

```python
def tise_solver(mass,potential,x_min, x_max, step_size):
 h0=0.1
 
 psi_nmin1=0
 psi_n=0.1

 psi=[psi_nmin1,psi_n]
 
 En=0.5
 Enmin1=0
 Enplus1=0
 n=(x_max-x_min)/step_size
 for i in range(0,n):
  h_bar_sq=np.square(6.582e-16)
  k_n_sq=2*mass*(En-potential)/h_bar_sq
  k_nmin1_sq=2*mass*(Enmin1-potential)/h_bar_sq
  k_nplus1_sq=2*mass*(Enplus1-potential)/h_bar_sq
  
  
  term1=1-(5/12 * (h0**2) * k_n_sq)
  term2=1+ (1/12 * (h0**2)* k_nmin1_sq)
  dr=1+(1/12)*(step_size*k_nplus1_sq)**2

  psi_nplus1=((2*term1*psi_n)-(term2*psi_nmin1))/dr
  
  psi_nmin1=psi_n
  psi_n=psi_nplus1

  psi.append(psi_nplus1)

 return psi
```
