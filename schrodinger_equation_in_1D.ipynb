```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def tise_solver(mass,potential,x_min, x_max, step_size,E):

 h_bar_sq=1
 psi_nmin1=0
 psi_n=0.1

 psi=[psi_nmin1,psi_n]
 
 
 #h_bar_sq=np.square(6.582e-16)
 n=int((x_max-x_min)/step_size)+1
 
 for i in range(1,n-1):
  
  k_n_sq=2*mass*(E-potential[i])/h_bar_sq
  k_nmin1_sq=2*mass*(E-potential[i-1])/h_bar_sq
  k_nplus1_sq=2*mass*(E-potential[i+1])/h_bar_sq
  
  
  term1=1-(5/12 * (step_size**2) * k_n_sq)
  term2=1+ (1/12 * (step_size**2)* k_nmin1_sq)
  dr=1+(1/12)*k_nplus1_sq*(step_size**2)

  psi_nplus1=((2*term1*psi_n)-(term2*psi_nmin1))/dr
  
  psi_nmin1=psi_n
  psi_n=psi_nplus1

  psi.append(psi_nplus1)

 return psi
```

```python
def qho(x_min, x_max, step_size, mass, omega, E):
    n=int((x_max-x_min)/step_size) +1
    x=np.linspace(x_min, x_max, n) 

    v= 0.5*mass* (omega**2) * (x**2)

    return v

```

```python
x_min=-5
x_max=5
step_size=0.01
n=int((x_max-x_min)/step_size) +1
mass=1
omega=1
E=.5
```

```python
potential_qho=qho(x_min, x_max, step_size, mass, omega, E )
psi_x=tise_solver(mass, potential_qho,x_min, x_max, step_size,E)
x=np.linspace(x_min, x_max, n)
```
